```python
route_mapping = bus_routes[
    ["route_id", "supplier_id"]
]
merged_data = journey_records.merge(
    route_mapping,
    on="route_id",
    how="left",
    suffixes=("_journey", "_route")
)
mismatches = merged_data[
    merged_data["supplier_id_journey"]
    != merged_data["supplier_id_route"]
]
print("SUPPLIER-ROUTE MAPPING")
print("Total mismatches:", len(mismatches))
print("\nQUANTITY VALIDATION")
print("Zero or negative ordered quantities:",
      (journey_records["quantity_ordered"] <= 0).sum())
print("Negative delivered quantities:",
      (journey_records["quantity_delivered"] < 0).sum())
print("\nMETRIC VALIDATION")
print("OTIF unique values:",
      sorted(journey_records["otif"].unique()))
print("Defect rate range:",
      journey_records["defect_rate"].min(),
      "-",
      journey_records["defect_rate"].max())
print("\nDATE COVERAGE")
for col in [
    "order_date",
    "committed_delivery_date",
    "actual_delivery_date"
]:
    dates = pd.to_datetime(journey_records[col])
    print(col, ":", dates.min(), "to", dates.max())
```

```python
expected_otif = (
    (journey_records["is_on_time"] == 1) &
    (journey_records["is_in_full"] == 1)
).astype(int)
otif_mismatches = (
    journey_records["otif"] != expected_otif
).sum()
print("OTIF mismatches:", otif_mismatches)
calculated_lead_time = (
    pd.to_datetime(journey_records["actual_delivery_date"]) -
    pd.to_datetime(journey_records["order_date"])
).dt.days
lead_time_mismatches = (
    journey_records["lead_time_days"] != calculated_lead_time
).sum()
print("Lead time mismatches:", lead_time_mismatches)
calculated_variance = (
    pd.to_datetime(journey_records["actual_delivery_date"]) -
    pd.to_datetime(journey_records["committed_delivery_date"])
).dt.days
variance_mismatches = (
    journey_records["variance_days"] != calculated_variance
).sum()
print("Variance mismatches:", variance_mismatches)
```

```python
lead_time_check = journey_records.copy()
lead_time_check["calculated_lead_time"] = (
    pd.to_datetime(lead_time_check["actual_delivery_date"]) -
    pd.to_datetime(lead_time_check["order_date"])
).dt.days
mismatched_records = lead_time_check[
    lead_time_check["lead_time_days"] !=
    lead_time_check["calculated_lead_time"]
]
print("Mismatched Records:")
display(
    mismatched_records[
        [
            "record_id",
            "po_number",
            "order_date",
            "actual_delivery_date",
            "lead_time_days",
            "calculated_lead_time"
        ]
    ]
)
```

```python
summary_check = ridership_summary.copy()
summary_check["calculated_otif_rate"] = (
    summary_check["otif_count"] /
    summary_check["total_shipments"] * 100
).round(1)
summary_check["otif_difference"] = (
    summary_check["otif_rate"] -
    summary_check["calculated_otif_rate"]
).round(1)
print("OTIF Rate Mismatches:")
display(
    summary_check[
        summary_check["otif_difference"] != 0
    ][
        [
            "supplier_id",
            "month_year",
            "otif_rate",
            "calculated_otif_rate",
            "otif_difference"
        ]
    ]
)
print(
    "Total OTIF mismatches:",
    (summary_check["otif_difference"] != 0).sum()
)
```

```python
from decimal import Decimal, ROUND_HALF_UP
mismatch_rows = summary_check[
    summary_check["otif_difference"] != 0
]
for _, row in mismatch_rows.iterrows():
    exact_rate = (
        Decimal(str(row["otif_count"])) /
        Decimal(str(row["total_shipments"])) * 100
    )
    standard_rate = exact_rate.quantize(
        Decimal("0.1"),
        rounding=ROUND_HALF_UP
    )
    print(f"\nSupplier: {row['supplier_id']}")
    print(f"Month: {row['month_year']}")
    print(f"Exact OTIF Rate: {exact_rate}")
    print(f"Standard Rounded Rate: {standard_rate}")
    print(f"Dataset OTIF Rate: {row['otif_rate']}")
```

```python
from decimal import Decimal, ROUND_HALF_UP
def calculate_defect_rate(row):
    rate = (
        Decimal(int(row["defect_count"])) /
        Decimal(int(row["total_shipments"])) * 100
    )
    return float(
        rate.quantize(
            Decimal("0.1"),
            rounding=ROUND_HALF_UP
        )
    )
summary_check["calculated_defect_rate"] = summary_check.apply(
    calculate_defect_rate,
    axis=1
)
summary_check["defect_difference"] = (
    summary_check["defect_rate"] -
    summary_check["calculated_defect_rate"]
).round(1)
defect_mismatches = summary_check[
    summary_check["defect_difference"] != 0
]
print("Total Defect Rate Mismatches:", len(defect_mismatches))
display(
    defect_mismatches[
        [
            "supplier_id",
            "month_year",
            "defect_rate",
            "calculated_defect_rate",
            "defect_difference"
        ]
    ]
)
```

```python
import pandas as pd
cleaned_data = journey_records.copy()
date_columns = [
    "order_date",
    "committed_delivery_date",
    "actual_delivery_date"
]
for col in date_columns:
    cleaned_data[col] = pd.to_datetime(cleaned_data[col])
cleaned_data["corrected_lead_time_days"] = (
    cleaned_data["actual_delivery_date"] -
    cleaned_data["order_date"]
).dt.days
corrections = (
    cleaned_data["lead_time_days"] !=
    cleaned_data["corrected_lead_time_days"]
).sum()
print("Total Records:", len(cleaned_data))
print("Lead Time Corrections:", corrections)
print("Missing Values:", cleaned_data.isnull().sum().sum())
print("Duplicate Records:", cleaned_data.duplicated().sum())
print("Corrected Average Lead Time:",
      round(cleaned_data["corrected_lead_time_days"].mean(), 2))
```

```python
def prepare_supply_chain_data(data):
    df = data.copy()
    date_columns = [
        "order_date",
        "committed_delivery_date",
        "actual_delivery_date"
    ]
    for col in date_columns:
        df[col] = pd.to_datetime(df[col])
    df = df.drop_duplicates()
    df["corrected_lead_time_days"] = (
        df["actual_delivery_date"] - df["order_date"]
    ).dt.days
    df["calculated_variance_days"] = (
        df["actual_delivery_date"] -
        df["committed_delivery_date"]
    ).dt.days
    df["calculated_otif"] = (
        (df["actual_delivery_date"] <= df["committed_delivery_date"]) &
        (df["quantity_delivered"] >= df["quantity_ordered"])
    ).astype(int)
    df["month_year"] = df["order_date"].dt.strftime("%b-%Y")
    return df
processed_data = prepare_supply_chain_data(journey_records)
print("Data Pipeline Executed Successfully!")
print("Original Shape:", journey_records.shape)
print("Processed Shape:", processed_data.shape)
print("Missing Values:", processed_data.isnull().sum().sum())
print("Duplicate Records:", processed_data.duplicated().sum())
display(processed_data.head())
```

```python
supplier_metrics = processed_data.groupby("supplier_id").agg(
    total_orders=("po_number", "count"),
    successful_otif=("calculated_otif", "sum"),
    otif_rate=("calculated_otif", "mean"),
    avg_defect_rate=("defect_rate", "mean"),
    avg_lead_time=("corrected_lead_time_days", "mean"),
    avg_variance=("calculated_variance_days", "mean")
).reset_index()
supplier_metrics["otif_rate"] *= 100
supplier_metrics["otif_status"] = supplier_metrics["otif_rate"].apply(
    lambda x: "Below Target" if x < 85 else "Target Met"
)
supplier_metrics = supplier_metrics.round(2)
supplier_metrics = supplier_metrics.sort_values(
    "otif_rate", ascending=False
).reset_index(drop=True)
display(supplier_metrics)
```

```python
validation_results = {
    "Record Count Preserved":
        len(processed_data) == len(journey_records),
    "No Missing Values":
        processed_data.isnull().sum().sum() == 0,
    "No Duplicate Records":
        processed_data.duplicated().sum() == 0,
    "OTIF Calculation Verified":
        (processed_data["otif"] ==
         processed_data["calculated_otif"]).all(),
    "Variance Calculation Verified":
        (processed_data["variance_days"] ==
         processed_data["calculated_variance_days"]).all(),
    "Supplier Order Count Verified":
        supplier_metrics["total_orders"].sum() == len(processed_data),
    "Supplier OTIF Count Verified":
        supplier_metrics["successful_otif"].sum() ==
        processed_data["calculated_otif"].sum(),
    "All Suppliers Included":
        supplier_metrics["supplier_id"].nunique() ==
        journey_records["supplier_id"].nunique()
}
for check, result in validation_results.items():
    print(f"{check}: {'PASS' if result else 'FAIL'}")
print("\nOverall Validation:",
      "PASS" if all(validation_results.values()) else "FAIL")
```

```python
datasets = {
    "Bus Routes": bus_routes,
    "Journey Records": journey_records,
    "Ridership Summary": ridership_summary
}
reliability_results = []
for name, data in datasets.items():
    total_cells = data.size
    missing_cells = data.isnull().sum().sum()
    duplicate_rows = data.duplicated().sum()
    reliability_results.append({
        "Dataset": name,
        "Total Records": len(data),
        "Missing Values": missing_cells,
        "Duplicate Records": duplicate_rows,
        "Completeness (%)": round(
            (1 - missing_cells / total_cells) * 100, 2
        ),
        "Uniqueness (%)": round(
            (1 - duplicate_rows / len(data)) * 100, 2
        )
    })
reliability_report = pd.DataFrame(reliability_results)
display(reliability_report)
```
