```python
import pandas as pd
employees = pd.DataFrame({
    "employee_id": [101, 102, 103, 104],
    "name": ["Arun", "Bala", "Charan", "Divya"],
    "dept_id": [1, 2, 1, 3]
})
departments = pd.DataFrame({
    "dept_id": [1, 2, 3],
    "department": ["IT", "HR", "Finance"]
})
print("Employees Table:")
display(employees)
print("Departments Table:")
display(departments)
```

```python
inner_join = pd.merge(
    employees,
    departments,
    on="dept_id",
    how="inner"
)
display(inner_join)
```

```python
left_join = pd.merge(
    employees,
    departments,
    on="dept_id",
    how="left"
)
display(left_join)
```

```python
right_join = pd.merge(
    employees,
    departments,
    on="dept_id",
    how="right"
)
display(right_join)
```

```python
outer_join = pd.merge(
    employees,
    departments,
    on="dept_id",
    how="outer"
)
display(outer_join)
```

```python
sales = pd.DataFrame({
    "Region": ["South", "North", "South", "North", "South"],
    "Product": ["Laptop", "Laptop", "Phone", "Phone", "Laptop"],
    "Sales": [50000, 40000, 20000, 30000, 10000]
})
display(sales)
```

```python
sales_summary = pd.pivot_table(
    sales,
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)
display(sales_summary)
```

```python
salary = pd.DataFrame({
    "Employee": ["Arun", "Bala"],
    "January": [20000, 25000],
    "February": [22000, 26000],
    "March": [21000, 27000]
})
display(salary)
```

```python
long_data = pd.melt(
    salary,
    id_vars=["Employee"],
    var_name="Month",
    value_name="Salary"
)
display(long_data)
```

```python
sales_data = pd.DataFrame({
    "Region": ["South", "North", "South", "West", "North"],
    "Sales": [15000, 8000, 25000, None, 30000]
})
result = (
    sales_data
    .dropna(subset=["Sales"])
    .query("Sales > 10000")
    .groupby("Region", as_index=False)["Sales"]
    .sum()
    .sort_values("Sales", ascending=False)
)
display(result)
```
