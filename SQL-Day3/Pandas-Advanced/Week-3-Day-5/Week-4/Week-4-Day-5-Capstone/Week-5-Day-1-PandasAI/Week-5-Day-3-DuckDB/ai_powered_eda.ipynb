```python
salary_issues = df_cleaned[
    df_cleaned["Monthly Salary"].isnull() &
    (df_cleaned["Salary"] != "Not specified")
]
print("Salary conversion issues:", len(salary_issues))
display(
    salary_issues[
        ["Job Title", "Location", "Salary", "Monthly Salary"]
    ].head(30)
)
print("\nExisting salary examples:")
display(
    df_cleaned[
        df_cleaned["Monthly Salary"].notnull()
    ][["Salary", "Monthly Salary"]].head(20)
)
```

```python
import re
import numpy as np
def convert_to_monthly_salary(salary):
    if pd.isna(salary) or salary == "Not specified":
        return np.nan
    amounts = re.findall(r'[\d,]+(?:\.\d+)?', salary)
    if not amounts:
        return np.nan
    values = [float(x.replace(",", "")) for x in amounts]
    if len(values) >= 2:
        amount = sum(values[:2]) / 2
    else:
        amount = values[0]
    if "year" in salary.lower():
        amount = amount / 12
    return amount
df_cleaned["Monthly Salary Cleaned"] = (
    df_cleaned["Salary"].apply(convert_to_monthly_salary)
)
print("Salary conversion completed!")
display(
    df_cleaned[
        ["Salary", "Monthly Salary", "Monthly Salary Cleaned"]
    ].head(20)
)
print("\nMissing values after conversion:")
print(df_cleaned["Monthly Salary Cleaned"].isnull().sum())
```

```python
df_cleaned["Locality Cleaned"] = df_cleaned["Locality"]
def extract_locality(location):
    if pd.isna(location):
        return np.nan
    parts = [part.strip() for part in location.split(",")]
    return parts[0]
missing_locality = df_cleaned["Locality Cleaned"].isnull()
df_cleaned.loc[missing_locality, "Locality Cleaned"] = (
    df_cleaned.loc[missing_locality, "Location"].apply(extract_locality)
)
print("Locality cleaning completed!")
print("\nMissing Locality Before:", df_cleaned["Locality"].isnull().sum())
print("Missing Locality After:", df_cleaned["Locality Cleaned"].isnull().sum())
display(
    df_cleaned[
        ["Location", "Locality", "Locality Cleaned"]
    ].head(25)
)
```

```python
df_cleaned["Final Monthly Salary"] = df_cleaned["Monthly Salary Cleaned"]
print("Salary Standardization Completed!")
print("\nSalary Summary:")
display(df_cleaned["Final Monthly Salary"].describe())
print("\nMissing Salary Values:")
print(df_cleaned["Final Monthly Salary"].isnull().sum())
print("\nSample Records:")
display(
    df_cleaned[
        ["Job Title", "Salary", "Final Monthly Salary"]
    ].head(20)
)
```

```python
print("Lowest 15 Salaries:")
display(
    df_cleaned.nsmallest(
        15, "Final Monthly Salary"
    )[["Job Title", "Salary", "Final Monthly Salary"]]
)
print("\nHighest 15 Salaries:")
display(
    df_cleaned.nlargest(
        15, "Final Monthly Salary"
    )[["Job Title", "Salary", "Final Monthly Salary"]]
)
```

```python
def convert_to_monthly_salary(salary):
    if pd.isna(salary) or salary == "Not specified":
        return np.nan
    amounts = re.findall(r'[\d,]+(?:\.\d+)?', salary)
    if not amounts:
        return np.nan
    values = [float(x.replace(",", "")) for x in amounts]
    if len(values) >= 2:
        amount = sum(values[:2]) / 2
    else:
        amount = values[0]
    salary_lower = salary.lower()
    if "year" in salary_lower:
        return amount / 12
    elif "month" in salary_lower:
        return amount
    else:
        return np.nan
df_cleaned["Final Monthly Salary"] = (
    df_cleaned["Salary"].apply(convert_to_monthly_salary)
)
print("Corrected salary conversion completed!")
print("\nRemaining missing salaries:")
print(df_cleaned["Final Monthly Salary"].isnull().sum())
display(
    df_cleaned[
        ["Job Title", "Salary", "Final Monthly Salary"]
    ].head(20)
)
```

```python
print("Lowest 10 Monthly Salaries:")
display(
    df_cleaned.nsmallest(
        10, "Final Monthly Salary"
    )[["Job Title", "Salary", "Final Monthly Salary"]]
)
print("\nHighest 10 Monthly Salaries:")
display(
    df_cleaned.nlargest(
        10, "Final Monthly Salary"
    )[["Job Title", "Salary", "Final Monthly Salary"]]
)
print("\nUpdated Salary Statistics:")
display(df_cleaned["Final Monthly Salary"].describe())
```

```python
salary = df_cleaned["Final Monthly Salary"].dropna()
Q1 = salary.quantile(0.25)
Q3 = salary.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = salary[
    (salary < lower_bound) |
    (salary > upper_bound)
]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Total Outliers:", len(outliers))
print("Outlier Percentage:", round(len(outliers) / len(salary) * 100, 2), "%")
```

```python
state_salary = (
    df_cleaned
    .dropna(subset=["Final Monthly Salary"])
    .groupby("State")["Final Monthly Salary"]
    .agg(["mean", "median", "count"])
    .sort_values("mean", ascending=False)
)
state_salary.columns = [
    "Average Salary",
    "Median Salary",
    "Job Count"
]
display(state_salary)
```

```python
top_states = (
    state_salary[state_salary["Job Count"] >= 10]
    .sort_values("Job Count", ascending=False)
    .head(10)
    .sort_values("Average Salary", ascending=False)
)
plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_states.reset_index(),
    x="Average Salary",
    y="State"
)
plt.title("Average Monthly Salary – Top 10 States by Job Count")
plt.xlabel("Average Monthly Salary (INR)")
plt.ylabel("State")
plt.show()
```

```python
top_jobs = df_cleaned["Job Title"].value_counts().head(15)
print("Top 15 Most In-Demand Job Roles:")
display(top_jobs)
plt.figure(figsize=(12, 7))
sns.barplot(
    x=top_jobs.values,
    y=top_jobs.index
)
plt.title("Top 15 Most In-Demand Job Roles in India")
plt.xlabel("Number of Job Listings")
plt.ylabel("Job Title")
plt.show()
```

```python
job_salary = (
    df_cleaned
    .dropna(subset=["Final Monthly Salary"])
    .groupby("Job Title")["Final Monthly Salary"]
    .agg(["mean", "count"])
)
job_salary.columns = ["Average Salary", "Job Count"]
top_paying_jobs = (
    job_salary[job_salary["Job Count"] >= 2]
    .sort_values("Average Salary", ascending=False)
    .head(15)
)
display(top_paying_jobs)
plt.figure(figsize=(12, 7))
sns.barplot(
    data=top_paying_jobs.reset_index(),
    x="Average Salary",
    y="Job Title"
)
plt.title("Top 15 Highest Paying Job Roles")
plt.xlabel("Average Monthly Salary (INR)")
plt.ylabel("Job Title")
plt.show()
```
