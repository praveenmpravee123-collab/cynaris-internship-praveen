```python
from typing import Optional
from pydantic import Field
from vanna.tools.agent_memory import SearchSavedCorrectToolUsesParams

class LowThresholdSearchParams(SearchSavedCorrectToolUsesParams):
    similarity_threshold: Optional[float] = Field(
        default=0.3,
        description="Minimum similarity score for results"
    )

class LowThresholdSearchTool(SearchSavedCorrectToolUsesTool):

    def get_args_schema(self):
        return LowThresholdSearchParams

print("Custom Memory Search Tool created!")
print(
    "Default threshold:",
    LowThresholdSearchParams.model_fields["similarity_threshold"].default
)
```

```python
async def search_sql_examples(question):
    search_tool = StrictSearchTool()

    args = search_tool.get_args_schema()(
        question=question
    )

    result = await search_tool.execute(
        context=context,
        args=args
    )

    return result


async def execute_verified_sql(sql):
    args = db_tool.get_args_schema()(sql=sql)

    result = await db_tool.execute(
        context=context,
        args=args
    )

    return result
```

```python
question = "What is the maximum monthly salary?"

result = await search_sql_examples(question)

print(result)
```

```python
sql = result.result_for_llm.split("Args: ")[1]

import ast

sql_args = ast.literal_eval(sql)
verified_result = await execute_verified_sql(sql_args["sql"])

print(verified_result)
```

```python
async def ask_sql_question(question):
    print("Question:", question)

    # Step 1: Search saved SQL examples
    memory_result = await search_sql_examples(question)

    if not memory_result.success:
        print("Memory search failed:", memory_result.error)
        return

    print("\nMemory Search:")
    print(memory_result.result_for_llm)

    # Step 2: Extract saved SQL
    if "Args: " not in memory_result.result_for_llm:
        print("\nNo matching SQL example found.")
        return

    import ast

    saved_args = ast.literal_eval(
        memory_result.result_for_llm.split("Args: ", 1)[1]
    )

    # Step 3: Execute SQL
    sql = saved_args["sql"]

    print("\nExecuting SQL:")
    print(sql)

    execution_result = await execute_verified_sql(sql)

    print("\nExecution Result:")
    print(execution_result)

    return execution_result
print("done")
```
