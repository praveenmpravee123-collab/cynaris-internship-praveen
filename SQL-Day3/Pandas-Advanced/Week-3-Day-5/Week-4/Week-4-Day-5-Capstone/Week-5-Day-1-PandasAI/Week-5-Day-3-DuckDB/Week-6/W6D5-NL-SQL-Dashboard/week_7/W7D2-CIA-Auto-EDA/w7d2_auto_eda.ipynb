## AutoViz vs SweetViz Comparison

AutoViz and SweetViz are useful Python libraries for automated Exploratory Data Analysis. AutoViz generates visualizations such as distribution plots, violin plots, and categorical charts, helping analysts quickly identify patterns and outliers. SweetViz provides a detailed HTML report containing dataset summaries, missing values, feature distributions, and associations. In this project, AutoViz generated three HTML chart reports, while SweetViz produced one comprehensive report covering all six dataset features. AutoViz is useful for quick visual exploration, whereas SweetViz offers a structured overview that is easier to share with business stakeholders. Both tools complement each other in understanding dataset quality, patterns and insights.

## Business Stakeholder Recommendation

I recommend SweetViz for business stakeholders because it
provides a comprehensive and structured HTML report.
It presents missing values, feature distributions, and
dataset summaries in one place, making the findings
easier to review and communicate.

AutoViz is useful for analysts who need quick,
interactive visual exploration. Both tools can be used
together for effective exploratory data analysis.
