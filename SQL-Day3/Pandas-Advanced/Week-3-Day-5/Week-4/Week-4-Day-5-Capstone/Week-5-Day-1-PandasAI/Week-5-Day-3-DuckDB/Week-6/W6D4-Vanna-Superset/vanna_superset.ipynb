```python
from vanna.core.registry import ToolRegistry
from vanna.core.user import UserResolver, RequestContext, User
from vanna.core.tool.models import ToolContext
from vanna.core.system_prompt.base import SystemPromptBuilder
from vanna.core.system_prompt.default import DefaultSystemPromptBuilder
from vanna.integrations.local.agent_memory import DemoAgentMemory
from vanna.tools import RunSqlTool

agent_memory = DemoAgentMemory(max_items=10000)

tools = ToolRegistry()

sql_tool = RunSqlTool(
    sql_runner=sql_runner
)

tools.register_local_tool(
    sql_tool,
    access_groups=["admin", "user"]
)

print("Agent Memory: Configured")
print("Tool Registry: Created")
print("SQL Tool: Registered")
```

```python
class SimpleUserResolver(UserResolver):

    async def resolve_user(self, request_context: RequestContext) -> User:
        return User(
            id="sales_analyst",
            username="Sales Analyst",
            group_memberships=["admin", "user"]
        )


user_resolver = SimpleUserResolver()

print("User Resolver: Configured")
print("User ID: sales_analyst")
print("Access Groups: admin, user")
```

```python
class SalesSystemPromptBuilder(SystemPromptBuilder):

    def __init__(self):
        self.default_builder = DefaultSystemPromptBuilder()

    async def build_system_prompt(self, user, tools):
        base_prompt = await self.default_builder.build_system_prompt(
            user=user,
            tools=tools
        )

        schema_instructions = """
DATABASE SCHEMA INSTRUCTIONS:

Database: practice.db
Table name: sales

Available columns:
- id (INTEGER)
- product (TEXT)
- category (TEXT)
- quantity (INTEGER)
- price (REAL)
- sales_date (DATE)

STRICT SQL RULES:
1. Always use the exact table name sales.
2. Use only the available columns.
3. Use SQLite-compatible SQL syntax.
4. Calculate revenue using quantity * price.
5. Always use run_sql for database questions.
6. Never invent columns or table names.
"""

        return f"{base_prompt}\n\n{schema_instructions}"


custom_prompt_builder = SalesSystemPromptBuilder()

print("Sales System Prompt: Configured")
print("Database Schema: Added")
```

```python
agent = Agent(
    llm_service=llm,
    tool_registry=tools,
    user_resolver=user_resolver,
    agent_memory=agent_memory,
    system_prompt_builder=custom_prompt_builder
)

print("Vanna Agent: Created Successfully")
print("LLM: qwen2.5:3b")
print("Database: practice.db")
print("Table: sales")
```

```python
TRAINING_EXAMPLES = [
    (
        "What is the total sales revenue?",
        "SELECT SUM(quantity * price) AS total_sales FROM sales;"
    ),
    (
        "What is the category-wise sales revenue?",
        "SELECT category, SUM(quantity * price) AS total_sales FROM sales GROUP BY category;"
    ),
    (
        "What are the top selling products by revenue?",
        "SELECT product, SUM(quantity * price) AS total_sales FROM sales GROUP BY product ORDER BY total_sales DESC LIMIT 5;"
    ),
    (
        "What is the monthly sales revenue?",
        "SELECT strftime('%Y-%m', sales_date) AS month, SUM(quantity * price) AS monthly_sales FROM sales GROUP BY month ORDER BY month;"
    ),
    (
        "How many units were sold in each category?",
        "SELECT category, SUM(quantity) AS total_quantity FROM sales GROUP BY category;"
    )
]

print("Total Training Examples:", len(TRAINING_EXAMPLES))

for i, (question, sql) in enumerate(TRAINING_EXAMPLES, 1):
    print(f"\nExample {i}:")
    print("Question:", question)
    print("SQL:", sql)
```

```python
import uuid

api_user = User(
    id="sales_analyst",
    username="Sales Analyst",
    group_memberships=["admin", "user"]
)

training_context = ToolContext(
    user=api_user,
    conversation_id=str(uuid.uuid4()),
    request_id=str(uuid.uuid4()),
    agent_memory=agent_memory
)

for question, sql in TRAINING_EXAMPLES:
    await agent_memory.save_tool_usage(
        question=question,
        tool_name="run_sql",
        args={"sql": sql},
        context=training_context,
        success=True
    )

print("Five training examples saved successfully!")
```

```python
print("Training Examples Verification")
print("=" * 40)

for i, (question, sql) in enumerate(TRAINING_EXAMPLES, 1):
    print(f"{i}. {question}")
    print(f"   SQL: {sql}")

print("\nTotal Examples Saved:", len(TRAINING_EXAMPLES))
```
