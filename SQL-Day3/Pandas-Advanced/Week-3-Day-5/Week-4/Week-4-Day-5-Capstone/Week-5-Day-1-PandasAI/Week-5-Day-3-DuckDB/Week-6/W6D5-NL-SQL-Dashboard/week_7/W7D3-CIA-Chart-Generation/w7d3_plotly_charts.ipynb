```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
print("Libraries imported successfully!")
```

```python
data = {
    "Month": ["January", "February", "March", "April", "May", "June"],
    "Revenue": [120000, 150000, 135000, 180000, 200000, 225000],
    "Orders": [120, 150, 135, 180, 200, 225],
    "Customers": [80, 95, 90, 110, 125, 140]
}
df = pd.DataFrame(data)
print("Dataset Created Successfully!")
display(df)
```

```python
fig = px.bar(
    df,
    x="Month",
    y="Revenue",
    title="Monthly Revenue Analysis",
    labels={
        "Month": "Month",
        "Revenue": "Revenue (₹)"
    },
    text="Revenue"
)
fig.update_traces(textposition="outside")
fig.update_layout(
    template="plotly_white",
    xaxis_title="Month",
    yaxis_title="Revenue (₹)"
)
fig.show()
```

```python
fig = px.line(
    df,
    x="Month",
    y="Revenue",
    title="Monthly Revenue Trend",
    markers=True,
    text="Revenue"
)
fig.update_traces(
    line=dict(width=3),
    marker=dict(size=10),
    textposition="top center"
)
fig.update_layout(
    template="plotly_white",
    xaxis_title="Month",
    yaxis_title="Revenue (₹)"
)
fig.show()
```

```python
fig = px.scatter(
    df,
    x="Orders",
    y="Customers",
    size="Revenue",
    color="Month",
    hover_name="Month",
    title="Orders vs Customers Analysis",
    size_max=40
)
fig.update_layout(
    template="plotly_white",
    xaxis_title="Total Orders",
    yaxis_title="Total Customers"
)
fig.show()
```

```python
fig = px.pie(
    df,
    names="Month",
    values="Revenue",
    title="Monthly Revenue Distribution",
    hole=0.4
)
fig.update_traces(
    textinfo="label+percent",
    hovertemplate="%{label}<br>Revenue: ₹%{value:,}<br>Share: %{percent}<extra></extra>"
)
fig.update_layout(
    template="plotly_white"
)
fig.show()
```

```python
# ECharts - Monthly Revenue Bar Chart

from IPython.display import HTML, display
import json

months = df["Month"].tolist()
revenues = df["Revenue"].tolist()

chart_data = json.dumps({
    "months": months,
    "revenues": revenues
})

html_code = f"""
<div id="revenue_chart" style="width:100%; height:450px;"></div>

<script src="https://cdn.jsdelivr.net/npm/echarts@5/dist/echarts.min.js"></script>

<script>
(function() {{
    const chart = echarts.init(
        document.getElementById('revenue_chart')
    );

    const data = {chart_data};

    const option = {{
        title: {{
            text: 'Monthly Revenue Analysis - ECharts'
        }},
        tooltip: {{
            trigger: 'axis'
        }},
        xAxis: {{
            type: 'category',
            data: data.months
        }},
        yAxis: {{
            type: 'value',
            name: 'Revenue (₹)'
        }},
        series: [{{
            name: 'Revenue',
            type: 'bar',
            data: data.revenues,
            itemStyle: {{
                color: '#5470C6'
            }}
        }}]
    }};

    chart.setOption(option);
}})();
</script>
"""

display(HTML(html_code))
```

```python
from pathlib import Path

html_page = f"""
<!DOCTYPE html>
<html>
<head>
    <title>ECharts Monthly Revenue</title>
    <script src="https://cdn.jsdelivr.net/npm/echarts@5/dist/echarts.min.js"></script>
</head>
<body>
    <div id="chart" style="width:100%;height:550px;"></div>

    <script>
        const chart = echarts.init(document.getElementById('chart'));

        chart.setOption({{
            title: {{
                text: 'Monthly Revenue Analysis - ECharts'
            }},
            tooltip: {{
                trigger: 'axis'
            }},
            xAxis: {{
                type: 'category',
                data: {json.dumps(months)}
            }},
            yAxis: {{
                type: 'value',
                name: 'Revenue (₹)'
            }},
            series: [{{
                name: 'Revenue',
                type: 'bar',
                data: {json.dumps(revenues)},
                label: {{
                    show: true,
                    position: 'top'
                }}
            }}]
        }});

        window.addEventListener('resize', () => chart.resize());
    </script>
</body>
</html>
"""

output_file = Path("echarts_monthly_revenue.html")
output_file.write_text(html_page, encoding="utf-8")

print("ECharts HTML file created successfully!")
print(output_file.resolve())
```

```python
import webbrowser

webbrowser.open(output_file.resolve().as_uri())
```

```python
# ECharts - Monthly Revenue Donut Chart

donut_html = f"""
<!DOCTYPE html>
<html>
<head>
    <title>ECharts Revenue Distribution</title>
    <script src="https://cdn.jsdelivr.net/npm/echarts@5/dist/echarts.min.js"></script>
</head>
<body>
    <div id="chart" style="width:100%;height:550px;"></div>

    <script>
        const chart = echarts.init(document.getElementById('chart'));

        chart.setOption({{
            title: {{
                text: 'Monthly Revenue Distribution'
            }},
            tooltip: {{
                trigger: 'item'
            }},
            series: [{{
                name: 'Revenue',
                type: 'pie',
                radius: ['40%', '70%'],
                data: {json.dumps([
                    {"name": m, "value": int(r)}
                    for m, r in zip(months, revenues)
                ])},
                label: {{
                    show: true,
                    formatter: '{{b}}: {{d}}%'
                }}
            }}]
        }});

        window.addEventListener('resize', () => chart.resize());
    </script>
</body>
</html>
"""

donut_file = Path("echarts_revenue_donut.html")
donut_file.write_text(donut_html, encoding="utf-8")

print("ECharts Donut Chart created successfully!")
```

```python
webbrowser.open(donut_file.resolve().as_uri())
```
