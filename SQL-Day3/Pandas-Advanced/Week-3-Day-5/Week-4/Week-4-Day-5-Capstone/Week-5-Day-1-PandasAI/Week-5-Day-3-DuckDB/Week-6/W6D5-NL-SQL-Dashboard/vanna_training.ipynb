```python
from vanna import Tool
from vanna.core.tool.models import ToolResult
from vanna.capabilities.sql_runner.models import RunSqlToolArgs

class CRMQueryTool(Tool):

    @property
    def name(self) -> str:
        return "run_sql"

    @property
    def description(self) -> str:
        return "Execute read-only SQL queries on the CRM customers database."

    def get_args_schema(self):
        return RunSqlToolArgs

    async def execute(self, context, args: RunSqlToolArgs) -> ToolResult:
        try:
            result = await sql_runner.run_sql(args, context)

            return ToolResult(
                success=True,
                result_for_llm=result.to_string(index=False)
            )

        except Exception as e:
            return ToolResult(
                success=False,
                result_for_llm=f"SQL execution failed: {e}",
                error=str(e)
            )

crm_sql_tool = CRMQueryTool()

print("CRM SQL Tool created successfully!")
print("Tool Name:", crm_sql_tool.name)
print("Tool Description:", crm_sql_tool.description)
```

```python
training_examples = [
    {
        "question": "How many customers are there in the CRM database?",
        "sql": "SELECT COUNT(*) AS customer_count FROM customers;"
    },
    {
        "question": "What is the total deal value of all customers?",
        "sql": "SELECT SUM(deal_value) AS total_deal_value FROM customers;"
    },
    {
        "question": "How many customers are there in each deal stage?",
        "sql": """
            SELECT deal_stage, COUNT(*) AS customer_count
            FROM customers
            GROUP BY deal_stage
            ORDER BY customer_count DESC;
        """
    },
    {
        "question": "Which lead sources generate the most customers?",
        "sql": """
            SELECT lead_source, COUNT(*) AS customer_count
            FROM customers
            GROUP BY lead_source
            ORDER BY customer_count DESC;
        """
    },
    {
        "question": "Who are the top 5 sales representatives by total deal value?",
        "sql": """
            SELECT sales_rep, SUM(deal_value) AS total_deal_value
            FROM customers
            GROUP BY sales_rep
            ORDER BY total_deal_value DESC
            LIMIT 5;
        """
    }
]

print("Total Training Examples:", len(training_examples))

for i, example in enumerate(training_examples, start=1):
    print(f"\n{i}. {example['question']}")
    print(example["sql"])
```

```python
for i, example in enumerate(training_examples, start=1):
    print(f"\n{'='*55}")
    print(f"TRAINING EXAMPLE {i}")
    print("Question:", example["question"])

    try:
        result = await sql_runner.run_sql(
            RunSqlToolArgs(sql=example["sql"]),
            context
        )
        print("\nSQL Result:")
        print(result)
        print("Status: SUCCESS")

    except Exception as e:
        print("Status: FAILED")
        print("Error:", e)
```
