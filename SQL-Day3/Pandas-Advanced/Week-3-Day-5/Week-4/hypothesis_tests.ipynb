```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
scenarios = {
    "Scenario 1: Marketing campaign":{
        "H0": "The new marketing campaign does not increase the average conversion rate.",
        "H1": "The new marketing campaign incrrease the average conversion rate."
    },
    "Scenerio 2: Website Performance":{
        "H0":"The new website design does not change the average time spent by useers.",
        "H1":"The new website design change the average time spent by users."
    },
    "Scenerio 3: Customer satisfaction":{
        "H0":"The average customer satisfaction score is equal to 4.",
        "H1":"The average customer satisfaction score is not equal to 4"
    }
}
for scenario, hypotheses in scenarios.items():
    print(f"\n{scenario}")
    print(f"null hypotheses(H0): {hypotheses['H0']}")
    print(f"Alternative hypotheses(H1): {hypotheses["H1"]}")
```

```python
customer_scores = [4.2, 3.8, 4.1, 3.9, 4.5,
                   4.0, 3.7, 4.3, 4.1, 3.9]
claimed_mean = 4.0
t_statistic, p_value = stats.ttest_1samp(
    customer_scores, popmean=claimed_mean
)
sample_mean = np.mean(customer_scores)
print("One-Sample t-Test Results")
print("-" * 30)
print(f"Claimed Mean: {claimed_mean}")
print(f"Sample Mean: {sample_mean:.2f}")
print(f"T-Statistic: {t_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print("\nDecision: Reject the null hypothesis (H0).")
    print("The average satisfaction score differs significantly from 4.0.")
else:
    print("\nDecision: Fail to reject the null hypothesis (H0).")
    print("There is insufficient evidence that the average satisfaction score differs from 4.0.")
```

```python
sales_before = [120, 135, 128, 140, 125, 132, 138, 130]
sales_after = [145, 150, 142, 155, 148, 160, 152, 147]
t_statistic,p_value = stats.ttest_ind(
    sales_before,
    sales_after,
    equal_var=False
)
mean_before = np.mean(sales_before)
mean_after = np.mean(sales_after)
print("Two sample ttest results")
print("-"*40)
print(f"average sales before : {mean_before:.2f}")
print(f"average sales after : {mean_after:.2f}")
print(f"T-statistic : {t_statistic:.4f}")
print(f"P-vales : {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print("\nDecision: Reject the null hypothesis (H0).")
    print("There is a statistically significant difference in average sales.")
else:
    print("\nDecision: Fail to reject the null hypothesis (H0).")
    print("There is insufficient evidence of a difference in average sales.")
```

```python
customer_scores = [4.2, 3.8, 4.1, 3.9, 4.5,
                   4.0, 3.7, 4.3, 4.1, 3.9]
claimed_mean = 4.0
alpha = 0.05
t_statistic,p_value = stats.ttest_1samp(
    customer_scores, popmean=claimed_mean
)
confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(customer_scores)-1,
    loc=np.mean(customer_scores),
    scale=stats.sem(customer_scores)
)
print("P-value and confidence interval")
print("-"*40)
print(f"p-value : {p_value:.4f}")
print(f"95% confidence interval : {confidence_interval}")
print(f"sample mean : {np.mean(customer_scores):.2f}")
if p_value < alpha:
    print("\nStatistically significant result. ")
else:
    print("\nNot statistically significant")
print("\n Interpretation :")
print("The confidence interval gives a range of plausible values")
print("for the population's average customers satisfaction score")
```
