```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
salary_data = pd.DataFrame({
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Salary": [20000, 22000, 25000, 25000, 28000,
               30000, 32000, 35000, 40000, 80000]
})
salary_data
```

```python
mean_salary = salary_data["Salary"].mean()
median_salary = salary_data["Salary"].median()
mode_salary = salary_data["Salary"].mode()
print("Mean Salary:", mean_salary)
print("Median Salary:", median_salary)
print("Mode Salary:", mode_salary.tolist())
```

```python
variance_salary = salary_data["Salary"].var()
std_salary = salary_data["Salary"].std()
print("Salary Variance:", variance_salary)
print("Salary Standard Deviation:", std_salary)
```

```python
frequency = pd.cut(salary_data["Salary"], bins=5).value_counts().sort_index()
print("Salary Frequency Distribution:")
print(frequency)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(salary_data["Salary"], bins=5, edgecolor="black")
plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=salary_data,x="Salary",bins=5,kde=True)
plt.title("Salary Distribution with KDE")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
percentiles = salary_data["Salary"].quantile([0.25, 0.50, 0.75, 0.90])
print("25th Percentile (Q1):", percentiles[0.25])
print("50th Percentile (Q2):", percentiles[0.50])
print("75th Percentile (Q3):", percentiles[0.75])
print("90th Percentile (P90):", percentiles[0.90])
```

```python
salary_data["Group"] = [
    "Group A", "Group A", "Group A", "Group A", "Group A",
    "Group B", "Group B", "Group B", "Group B", "Group B"
]
print(salary_data)
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=salary_data, x="Group", y="Salary")
plt.title("Salary Comparison Between Two Groups")
plt.xlabel("Employee Group")
plt.ylabel("Salary")
plt.show()
```

```python
group_summary = salary_data.groupby("Group")["Salary"].agg(
    ["mean", "median", "std", "min", "max"]
)
print("Salary Comparison Between Groups:")
display(group_summary)
```

```python
print("STATISTICAL ANALYSIS - KEY FINDINGS")
print("-" * 40)
print("1. Central Tendency:")
print(f"   Mean Salary: ₹{mean_salary:.2f}")
print(f"   Median Salary: ₹{median_salary:.2f}")
print(f"   Mode Salary: ₹{mode_salary.iloc[0]:.2f}")
print("\n2. Data Variability:")
print(f"   Variance: {variance_salary:.2f}")
print(f"   Standard Deviation: ₹{std_salary:.2f}")
print("\n3. Distribution:")
print("   The ₹80,000 salary is substantially higher than most other salaries.")
print("   This high value increases the mean and contributes to salary variability.")
print("\n4. Group Comparison:")
print("   Compare the group means, medians, and standard deviations in the summary table.")
print("\n5. Conclusion:")
print("   Median salary provides a useful measure of typical salary when extreme values are present.")
```

# Statistical Analysis — Findings and Conclusion

## 1. Objective
The objective of this analysis is to calculate descriptive statistics, understand salary distributions, analyse percentiles and quartiles, and compare two employee groups.

## 2. Descriptive Statistics
Mean, median, and mode were calculated to understand the central tendency of employee salaries. Variance and standard deviation were used to measure salary variability.

## 3. Distribution Analysis
A frequency distribution, histogram, and KDE plot were created to visualise the salary distribution. The dataset contains a high salary value of ₹80,000 compared with the other observations.

## 4. Percentiles and Quartiles
The 25th, 50th, and 75th percentiles were calculated to understand the spread of salaries. The 90th percentile was also calculated to examine the upper portion of the distribution.

## 5. Group Comparison
A box plot and a descriptive statistics summary were used to compare Group A and Group B. The comparison included mean, median, standard deviation, minimum, and maximum salary.

## 6. Key Findings
- Mean, median, and mode describe different aspects of salary distribution.
- The high salary value affects the mean and contributes to overall variability.
- Percentiles and quartiles help explain the distribution of salaries.
- Group-wise statistics make it easier to compare salary patterns.

## 7. Conclusion
Descriptive statistics and visualisations help analysts understand data distribution, identify unusual values for further investigation, and compare groups. Median, standard deviation, percentiles, and box plots provide complementary insights into the dataset.
