```python
numeric_columns = df.select_dtypes(include="number").columns
print("Numeric Columns:")
print(numeric_columns.tolist())
```

```python
for column in numeric_columns:
    plt.figure(figsize=(7, 4))
    sns.histplot(data=df, x=column, bins=20, kde=True)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
correlation_matrix = numeric_df.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap of Supermarket Sales")
plt.tight_layout()
plt.show()
```

```python
# 1. Overall sales evlo?
print("Total Sales:", round(df["Sales"].sum(), 2))

# 2. Endha branch highest sales generate pannudhu?
print("\nSales by Branch:")
print(df.groupby("Branch")["Sales"].sum().sort_values(ascending=False))

# 3. Endha product category highest sales generate pannudhu?
print("\nSales by Product Line:")
print(df.groupby("Product line")["Sales"].sum().sort_values(ascending=False))

# 4. Endha branch-ku higher customer rating irukku?
print("\nAverage Rating by Branch:")
print(df.groupby("Branch")["Rating"].mean().round(2))

# 5. Customers endha payment method adhigama use panraanga?
print("\nPayment Method Count:")
print(df["Payment"].value_counts())
```

```python
branch_sales = df.groupby("Branch")["Sales"].sum().sort_values(ascending=False)
branch_sales.plot(kind="bar", figsize=(6, 4))
plt.title("Total Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby("Product line")["Sales"].sum().sort_values(ascending=False)
product_sales.plot(kind="bar", figsize=(10, 5))
plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
payment_counts = df["Payment"].value_counts()
payment_counts.plot(kind="bar", figsize=(6, 4))
plt.title("Payment Method Usage")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
branch_rating = df.groupby("Branch")["Rating"].mean()
branch_rating.plot(kind="bar", figsize=(6, 4))
plt.title("Average Customer Rating by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
