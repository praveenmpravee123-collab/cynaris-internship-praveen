```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Prave\OneDrive\Documents\employees.csv")

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6, 4))
sns.set_theme(style="whitegrid", palette="deep")
sns.barplot(data=df, x="Department", y="Salary")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary (₹)")
plt.savefig("bar_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.set_theme(style="whitegrid", palette="deep")
sns.histplot(data=df, x="Salary", bins=5, kde=True)
plt.title("Employee Salary Distribution")
plt.xlabel("Salary (₹)")
plt.ylabel("Number of Employees")
plt.savefig("histogram.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid", palette="deep")
sns.scatterplot(data=df, x="Name", y="Salary", hue="Department", s=120)
plt.title("Employee Salary by Name and Department")
plt.xlabel("Employee Name")
plt.ylabel("Salary (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("scatter_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid", palette="deep")
sns.boxplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary (₹)")
plt.savefig("box_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid", palette="deep")
sns.lineplot(data=df, x="Name", y="Salary", marker="o")
plt.title("Employee Salary Comparison")
plt.xlabel("Employee Name")
plt.ylabel("Salary (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("line_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```
